# **Библиотека реалиционных баз данных (SQLite)**

## **Основные типы данных**

| Тип данных SQLite | Описание | Тип данных Python |
| :--- | :--- | :--- |
| `NULL` | Отсутствие значения. | `None` |
| `INTEGER` | Целое число. | `int` |
| `REAL` | Число с плавающей точкой. | `float` |
| `TEXT` | Текстовая строка. | `str` |
| `BLOB` | Бинарные данные (изображения, файлы и т.д.). | `bytes` |

## **Основные команды**

| Объект / Функция | Метод | Описание |
| :--- | :--- | :--- |
| `sqlite3` | `.connect(database)` | Устанавливает соединение с базой данных. Создает файл, если его нет. |
| `Connection` | `.cursor()` | Создает объект `Cursor` для выполнения запросов. |
| `Connection` | `.commit()` | Сохраняет (фиксирует) все изменения, сделанные в текущей транзакции. |
| `Connection` | `.rollback()` | Откатывает все изменения, сделанные в текущей транзакции. |
| `Connection` | `.close()` | Закрывает соединение с базой данных. |
| `Cursor` | `.execute(sql, parameters)` | Выполняет SQL-запрос. Может принимать параметры для безопасной подстановки значений. |
| `Cursor` | `.executemany(sql, seq_of_params)` | Выполняет один и тот же SQL-запрос для каждой последовательности параметров. |
| `Cursor` | `.executescript(sql_script)` | Выполняет несколько SQL-запросов, разделенных точкой с запятой. |
| `Cursor` | `.fetchone()` | Возвращает одну следующую строку результата запроса в виде кортежа. |
| `Cursor` | `.fetchmany(size)` | Возвращает список из следующих `size` строк результата. |
| `Cursor` | `.fetchall()` | Возвращает список всех оставшихся строк результата. |/
