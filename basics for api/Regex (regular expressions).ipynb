# Регулярные выражения

## `re.match(pattern, string)`

```python
import re

''' Проверяет подходит ли наш пример под шаблон (то есть начинается ли наша строка как в примере)'''

string='Hello, how are you?'
pattern=r'Hello'
print(re.match(pattern, string))

''' Пример если строка не начинается со слов '''

string='Hello, how are you?'
pattern=r'how are you?'
print(re.match(pattern, string))

''' Пример если строка не подходит под шаблон '''

print(re.match('abc', 'acccd'))
```

## `re.search(pattern, string)`

```python
''' Находит первое вхождение строки (паттерна) в тексте'''

string='Hello, how are you?'
pattern=r'how are you?'
print(re.search(pattern, string))
```

## `re.findall(pattern, string)`

Возвращает подстроки подходящие под шаблон

```python
''' Находит все строки подходящие под шаблон (не пересекающиеся)'''

string = 'abcabcabc'
pattern = 'abc'

print(re.findall(pattern, string))
```

## `re.sub(pattern, repl, string)`

```python
''' Заменяет строки подходящие под шаблон на то, что вы захотите. Возвращает замененные подстроки'''

string = 'abc abc abc'
pattern = 'abc'
print(re.sub(pattern,'1',string))
```

## `re.split(pattern, string, [maxsplit=0])`

```python
''' Разбивает строку по шаблону'''

string = 'Hello'
pattern = 'll'

print(re.split(pattern, string))
```

## Расширение списка строк

```python
''' Например нам нужно, чтобы подходили не только строки с одним именем, но и с другим'''
''' Работает для всех функций описанных выше'''
string = 'Hello, Arseniy!'
string1 = 'Hello, everyone!'
pattern = r'[Arseniy,everyone]'

print(f'{re.search(pattern, string)}\n')
print(f'{re.search(pattern, string1)}')
```

### **Flags**

Коротко

- **`re.I` / re.IGNORECASE** — нечувствительность к регистру
- **`re.M` / re.MULTILINE** — ^ и $ применяются к каждой строке
- **`re.S`/ re.DOTALL** — . включает перевод строки
- **`re.X` / re.VERBOSE** — удобочитаемый формат с комментариями
- **`re.A` / re.ASCII** — только ASCII-символы для \w,\d,\b и т. п.
- **`re.L` / re.LOCALE** — поведение в соответствии с локалью (редко используется)


**Длинный ответ:**


- **Флаг IGNORECASE (`re.I`)** -- Делает поиск нечувствительным к регистру (a и A считаются одинаковыми).

- **Флаг MULTILINE (`re.M`)** -- ^ соответствует началу строки, $ — концу строки, а не только началу/концу всей строки. Работает, если в тексте есть переносы строк. (С флагом re.M они работают для каждой отдельной строки внутри текста — то есть для позиций сразу после \n и сразу перед \n.)

- **Флаг DOTALL (`re.S`)** -- Точка . также матчится с символом перевода строки \n (по умолчанию . не матчится с \n).

- **Флаг VERBOSE (`re.X`)** -- Позволяет писать регулярные выражения с пробелами и комментариями для удобства. Пробелы и комментарии (#...) игнорируются.

- **Флаг ASCII (`re.A`)** -- \w, \W, \b, \B, \d, \D работают только с ASCII-символами (полезно при работе с бинарными/старым данными).

    работать в ASCII-режиме, то есть интерпретировать их как наборы символов в диапазоне ASCII:
    - \w → [A-Za-z0-9_]
    - \d → [0-9]
    - \b (граница слова) опирается на эти ASCII-символы для определения границы.


- **Флаг LOCALE (`re.L`)** -- Меняет поведение специальных последовательностей (\w, \b и т. п.) в соответствии с текущей локалью C/POSIX. Используется редко.

https://docs-python.ru/standart-library/modul-re-python/flagi-ispolzuemye-funktsii-re-compile/

#### **Примеры только для часто используемых флагов**

#### **IGNORECASE** (`re.I`)

```python
text = 'HelLo world!'

re.search('hello', text, re.I)
```

#### **MULTILINE** (`re.M`)

```python
text = 'bar\nfoo'

print(f' Строка выглядит вот так:\n{text}\n{'---'*18}\nБез re.M: {re.search(r'^foo', text)}')

print(f'Без re.M: {re.search(r'^foo', text, re.M)}')
```

#### **DOTALL** (`re.S`)

```python
text = 'a\nb'

print(f'Без флага:\n{re.search(r'a.b',text)}\n{'---'*18}')

print(f'С флагом:\n{re.search(r'a.b',text, re.S)}')
```

#### **ASCII (`re.A`)**

```python
'''Без флага'''

print(re.match(r"\w+", "тест"))

'''С флагом'''
print(re.match(r'\w+', 'тест', re.A))
```
