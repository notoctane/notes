# **Lxml**

## **Шпаргалка с командами**

| Команда | Описание | Пример |
|---------|----------|--------|
| `etree.parse('file.xml')` | Парсинг XML файла | `tree = etree.parse('data.xml')` |
| `etree.parse('file.html', parser=etree.HTMLParser())` | Парсинг HTML файла | `tree = etree.parse('page.html', parser=etree.HTMLParser())` |
| `etree.fromstring('<tag>text</tag>')` | Парсинг XML строки | `root = etree.fromstring('<book>Title</book>')` |
| `etree.HTML('<html>...</html>')` | Парсинг HTML строки | `root = etree.HTML('<div>Hello</div>')` |
| `element.tag` | Получить имя тега | `root.tag` |
| `element.text` | Получить текст внутри тега | `div.text` |
| `element.attrib` | Получить словарь атрибутов | `a.attrib` |
| `element.get('attr')` | Получить значение атрибута | `a.get('href')` |
| `element.set('attr', 'value')` | Установить атрибут | `div.set('class', 'main')` |
| `etree.Element('tag')` | Создать новый элемент | `new = etree.Element('div')` |
| `parent.append(child)` | Добавить дочерний элемент | `root.append(new_div)` |
| `etree.SubElement(parent, 'tag')` | Создать и добавить дочерний элемент | `sub = etree.SubElement(root, 'span')` |
| `parent.remove(child)` | Удалить дочерний элемент | `root.remove(old_div)` |
| `element.getparent()` | Получить родительский элемент | `div.getparent()` |
| `element.getchildren()` | Получить список дочерних элементов | `root.getchildren()` |
| `element.find('child')` | Найти первого потомка по тегу | `root.find('title')` |
| `element.findall('.//tag')` | Найти все элементы по относительному пути | `root.findall('.//a')` |
| `element.xpath('//tag')` | Поиск по XPath | `root.xpath('//div[@class="item"]')` |
| `element.cssselect('div.class')` | Поиск по CSS селектору | `root.cssselect('div.main > a')` |
| `etree.tostring(element, encoding='unicode')` | Преобразовать в строку | `etree.tostring(root, encoding='unicode')` |
| `etree.tostring(element, pretty_print=True)` | Преобразовать с форматированием | `etree.tostring(root, pretty_print=True)` |
| `etree.tostring(element, method='html')` | Преобразовать как HTML | `etree.tostring(root, method='html')` |
| `etree.dump(element)` | Вывести в консоль (для отладки) | `etree.dump(root)` |
| `deepcopy(element)` | Глубокое копирование элемента | `new = deepcopy(root)` |

## **Пример**

```python
import requests
from lxml import etree
```

```python
''' Пример парсинга стоимости цен популярных предметов cs2 в steam '''

''' Для начала сделаем запрос '''

url = 'https://steamcommunity.com/market/search' # Ссылка на ресурс
params = {'appid' : '730'} # Передадим id игры в steam
steam_request = requests.get(url, params=params) # Сделаем запрос

''' Преобразуем HTML в xml'''

parser = etree.HTMLParser()
root = etree.fromstring(steam_request.text, parser)

''' Найдем все имена и стоимости предметов'''

item_name = root.xpath('//span[@class="market_listing_item_name"]') # Поиск всех имен
item_price = root.xpath('//span[@class="normal_price"]') # Поиск всех стоимосей этих предметов

for name, price in zip(item_name, item_price):
  print(f'{name.text}: {price.text}')
```

## **XPath простыми словами**

### 1. Как писать путь к элементу

| Символ | Что значит | Простыми словами |
|--------|------------|------------------|
| `/` | Слэш | **Иди строго внутрь.** Только к детям (первый уровень вложенности). |
| `//` | Два слэша | **Иди куда угодно.** Найдёшь на любом уровне, хоть глубоко внутри. |
| `.` | Точка | **Я сам.** Текущий элемент, на котором мы стоим. |
| `..` | Две точки | **Мой родитель.** Элемент, который содержит меня. |
| `@` | Собачка | **Атрибут.** Дополнительное свойство тега (например, class, id, href). |

---

### 2. Как фильтровать (предикаты) — квадратные скобки `[]`

| Конструкция | Что значит | Простыми словами |
|-------------|------------|------------------|
| `[1]` | Первый по счёту | Возьми первый элемент из найденных. |
| `[last()]` | Последний | Возьми последний элемент. |
| `[@class]` | У которого есть атрибут class | Возьми только те теги, у которых вообще есть class. |
| `[@class="name"]` | У которого class равен "name" | Возьми только те, у которых class именно такой. |
| `[contains(@class, "name")]` | У которого в class есть "name" | Возьми те, у которых внутри class встречается это слово. |
| `[text()="текст"]` | У которого текст равен "текст" | Возьми те, внутри которых написано именно это. |

---

### 3. Как искать по-разному (оси)

| Направление | Простыми словами |
|-------------|------------------|
| `child::` | Дети — теги, которые прямо внутри меня. |
| `parent::` | Родитель — тот, внутри которого я нахожусь. |
| `descendant::` | Все потомки — дети, внуки, правнуки и т.д. |
| `ancestor::` | Все предки — родитель, дедушка, корень. |
| `following-sibling::` | Братья и сёстры после меня — теги на том же уровне, которые идут позже. |
| `preceding-sibling::` | Братья и сёстры до меня — теги на том же уровне, которые идут раньше. |

---

### 4. Что такое оси в сокращённом виде

| Сокращение | Полная форма | Простыми словами |
|------------|--------------|------------------|
| `//` | `descendant-or-self::` | Я сам или кто-то у меня внутри, на любой глубине. |
| `@` | `attribute::` | Атрибут тега (не сам тег, а его свойство). |
| `.` | `self::` | Я сам. |
| `..` | `parent::` | Мой родитель. |

---

### 5. Самые полезные функции

| Функция | Что делает | Простыми словами |
|---------|------------|------------------|
| `text()` | Берёт текст внутри тега | Всё, что написано между открывающим и закрывающим тегом. |
| `contains()` | Проверяет, есть ли кусочек текста | Ищет частичное совпадение, не обязательно точное. |
| `starts-with()` | Проверяет, с чего начинается текст | Начинается ли строка с определённых букв. |
| `normalize-space()` | Убирает лишние пробелы | Очищает текст от пробелов в начале, конце и лишних внутри. |
| `position()` | Номер элемента по порядку | Какой по счёту этот элемент среди похожих. |
| `last()` | Номер последнего элемента | Узнать, сколько всего элементов, или взять последний. |

---

### 6. Логические связки в фильтрах

| Связка | Что значит | Простыми словами |
|--------|------------|------------------|
| `and` | И | Должны выполняться оба условия сразу. |
| `or` | ИЛИ | Достаточно хотя бы одного условия. |
| `not()` | НЕ | Всё, кроме того, что внутри скобок. |

---

### 7. Звёздочка — универсальный символ

| Символ | Что значит | Простыми словами |
|--------|------------|------------------|
| `*` | Любой тег | Всё равно какой тег, лишь бы подходил по условию. |
