# **CSV**

**CSV** (*Comma-Separated Values* — «значения, разделённые запятыми») — это простейший формат для хранения табличных данных, где каждая строка — это запись, а значения внутри строки разделены запятыми. Удобен для обмена данными между таблицами и базами данных.

## Основные правила CSV:
- **Разделитель** — значения в строке разделяются запятыми `,` (Но разделитель также можно задать самим)
- **Строки** — каждая новая запись начинается с новой строки
- **Заголовки** — первая строка может содержать названия столбцов (но это необязательно)
- **Кавычки** — если значение содержит запятую, перевод строки или кавычку, оно заключается в двойные кавычки `"`
- **Экранирование** — двойная кавычка внутри значения заменяется на две кавычки `""`
- **Пробелы** — пробелы вокруг разделителя обычно игнорируются (считаются частью значения)

https://docs.python.org/3/library/csv.html

```python
import csv

''' Запись '''
with open('data.csv', 'w') as f:
    writer = csv.writer(f, quoting=csv.QUOTE_MINIMAL)  # другие режимы: QUOTE_ALL, QUOTE_NONE
    writer.writerow(['id', 'name', 'note'])
    writer.writerow([1, 'Иванов, И.', 'Текст с запятой и "кавычками"'])
```

**quoting** управляет тем, когда значения (поля) будут обёрнуты в кавычки при записи CSV.
Режимы:

- `csv.QUOTE_MINIMAL` -- оборачивает в кавычки только те поля, в которых это необходимо (например, если поле содержит разделитель, перенос строки или символ кавычек).
    -  **Когда полезно:** большинство случаев — компактный и корректный CSV.
- `csv.QUOTE_ALL` -- оборачивает в кавычки все поля, даже если в них нет специальных символов.
    - **Когда полезно:** когда нужно гарантировать одинаковый формат для всех полей или для совместимости с некоторыми парсерами.
- `csv.QUOTE_NONNUMERIC` -- оборачивает в кавычки все нечисловые поля; при чтении csv.reader также попытается преобразовать такие поля в float.
    - **полезно**, если вы хотите явно отметить числа без кавычек, а всё остальное — в кавычках. На чтении это может изменить типы данных. 
- `csv.QUOTE_NONE` -- не оборачивает поля в кавычки вообще. Если поле содержит разделитель или перенос строки, возникнет ошибка или потребуется экранирование.
    - **Требования:** обычно вместе с этим нужно задать параметр escapechar (например, '\') чтобы экранировать специальные символы.
    - **Когда полезно:** редко; когда требуется полностью безкавычный формат с экранированием спецсимволов.

```python
''' Чтение '''
with open('data.csv') as f:
    reader = csv.reader(f, delimiter=',')  # Можно указать delimiter=';' -- по чему разделить,
    for row in reader:
        print(row)
```

```python
''' Чтение в словари (если есть заголовок) '''
with open('data.csv') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

```python
from collections import Counter 
import csv

primary_type = []
with open('Crimes.csv') as f:
    reader = csv.DictReader(f)
    for row in reader:
        primary_type.append(row['Primary Type'])
    print(Counter(primary_type).most_common(1))
```

# **JSON**

**JSON (JavaScript Object Notation)** — это общепринятый стандарт для обмена данными в интернете. Он стал настолько популярным благодаря своей простоте и эффективности. Структура json напоминает словарь в питоне:


**Сериализация** — это процесс преобразования структуры данных Python (например, словаря или списка) в текстовую строку формата JSON.

**Десериализация** — это обратный процесс: преобразование JSON-строки в объект Python.

## **Отличия JSON от dict**

| Характеристика | Python (dict) | JSON (JavaScript Object Notation) |
| :--- | :--- | :--- |
| **Булевы значения** | `True` / `False` | `true` / `false` |
| **Пустое значение** | `None` | `null` |
| **Ключи** | Могут быть разных типов (int, tuple, str и др.) | Всегда строки в двойных кавычках (`"key"`) |
| **Кавычки** | Можно использовать `' '` или `" "` | Только двойные кавычки `" "` для строк и ключей |
| **Кодировка (Юникод)** | Русские буквы отображаются как есть (например, `"Привет"`) | По умолчанию экранируются (`\u041f\u0440\u0438\u0432\u0435\u0442`)<br>Отключается параметром `ensure_ascii=False` |
| **Комментарии** | Поддерживаются (`# комментарий`) | Не поддерживаются |
| **Методы** | Имеет методы (`.keys()`, `.values()` и др.) | Это просто текст, методов не имеет |
| **Синтаксис** | Более свободный | Строгий синтаксис (лишние запятые в конце не допускаются) |

## **Сериализация** `json.dumps()`

**Сериализация** — это процесс преобразования структуры данных Python (например, словаря или списка) в текстовую строку формата JSON.

```python
import json

request = {'name' : 'Арсений',
          'age' : 20,
          'Man' : True,
          'School' : None}

jsons = json.dumps(request)
print(f'Python:\n{request}\n{'---'*20}\nJSON:\n{jsons}')
```

### Различные параметры `json.dumps()`:

- `indent`: задает количество пробелов для отступа на каждом уровне вложенности.

- `sort_keys`: если True, ключи в объектах (словарях) будут отсортированы по алфавиту.

- `ensure_ascii`: если False, символы Unicode (например, русские буквы) не экранируются и выводятся как есть.

- `separators`: кортеж из двух строк (item_separator, key_separator). По умолчанию `(', ', ': ')`. Для минимизации JSON можно указать `(',', ':')`.

- `skipkeys`: если True, пропускает ключи словаря с некорректными типами (например, `None`, `set`, функции) вместо выброса `TypeError`.

- `default`: функция, которая вызывается для объектов, не поддерживаемых JSON по умолчанию (например, `datetime`, `Decimal`).

- `check_circular`: если True (по умолчанию), проверяет циклические ссылки в объектах. Если обнаружены, выбрасывает исключение.

- `allow_nan`: если True (по умолчанию), позволяет использовать `NaN`, `Infinity`, `-Infinity` в JSON (в JavaScript они будут представлены как `null`). Если False, выбрасывает `ValueError`.

- `cls`: пользовательский класс-кодировщик (наследник `JSONEncoder`) для кастомной сериализации.

```python
json_with_parametrs = json.dumps(request, indent = 4, sort_keys=True, ensure_ascii=False)
print(json_with_parametrs)
```

## **Десериализация** `json.loads()`

**Десериализация** — это обратный процесс: преобразование JSON-строки в объект Python.

```python
string_json = json.loads('{"name" : "Арсений", "age" : 20, "Man" : true, "School" : null}')
string_json
```

### Различные параметры `json.loads()`:

- `encoding`: кодировка строки. В современных версиях Python игнорируется, так как строка уже должна быть декодирована в Unicode.

- `cls`: пользовательский класс-декодировщик (наследник `JSONDecoder`) для кастомной десериализации.

- `object_hook`: функция, которая вызывается для каждого декодированного объекта (словаря). Может использоваться для преобразования словаря в объект пользовательского класса.

- `parse_float`: функция для парсинга чисел с плавающей точкой. По умолчанию `float`. Можно указать, например, `Decimal`.

- `parse_int`: функция для парсинга целых чисел. По умолчанию `int`. Можно указать, например, `float`.

- `parse_constant`: функция для парсинга констант (`-Infinity`, `Infinity`, `NaN`). По умолчанию преобразует в соответствующие значения Python.

- `object_pairs_hook`: функция, которая вызывается для каждого декодированного объекта с упорядоченным списком пар (ключ, значение). Позволяет сохранить порядок ключей или преобразовать в `OrderedDict`.

## **Работа с файлами `json.load()` и `json.dump()`**

```python
'''Запись json-объекта в файл'''

with open('profile.json', "w") as f:
    json.dump(string_json, f, indent=4, ensure_ascii=False)
```

```python
'''Преобразование json-объекта в python-объект'''

with open('profile.json', 'r') as f:
    profile = json.load(f)
print(profile)
```

```python
import json

a = json.loads(str(input()))
a = {i['name'] : i['parents'] for i in a} 

def parents(classes):
    answer = []
    for parent in classes.keys():
        visited, parent = set(), [parent]
        while parent:
            vertex = parent.pop()
            if vertex not in visited:
                visited.add(vertex)
                parent.extend(set(classes[vertex])-visited)
        answer += visited
    for key in sorted(classes.keys()):
        print(key, ":", answer.count(key))
    return
parents(a)
```
