# **XML** (*eXtensible Markup Language или "расширяемый язык разметки"*)

**XML** - это теговый язык разметки, предназначенный для хранения и передачи данных в структурированном виде. В отличие от HTML, теги здесь не предопределены — их можно придумывать самостоятельно под свои задачи.

## **Основные принципы XML:**

1. **Хранение данных**  
   XML используется исключительно для структурированного хранения и обмена информацией между программами, а не для отображения на экране.

2. **Расширяемость**  
   Вы сами создаёте названия тегов, которые отражают смысл данных (например, `<книга>`, `<цена>`, `<адрес>`).

3. **Строгий синтаксис**  
   - У каждого открывающего тега обязательно есть закрывающий (`<тег>...</тег>`).
   - Документ содержит ровно один корневой элемент.
   - Теги чувствительны к регистру (`<Book>` и `<book>` — разные).

4. **Атрибуты**  
   Внутри открывающего тега можно добавлять атрибуты — дополнительные характеристики элемента. Они всегда записываются в формате `название="значение"` (значения строго в кавычках).  
   *Пример: `<цена валюта="RUB" соскидка="нет">500</цена>`*

5. **Иерархичность**  
   Данные организованы в виде дерева: один элемент может содержать другие вложенные элементы.

6. **Независимость от платформы**  
   XML-файл — это обычный текст, поэтому его можно читать и создавать в любой операционной системе и на любом языке программирования.

## **Использование библиотеки** `xml.etree.ElementTree`

```python
''' Импорт библиотеки '''

from xml.etree import ElementTree
```

### **Создание XML**


| Команда | Описание | Пример |
|---------|----------|--------|
| `ET.Element(tag)` | Создает новый элемент с указанным тегом | `root = ET.Element("studentlist")` |
| `ET.SubElement(parent, tag)` | Создает дочерний элемент внутри родительского | `student = ET.SubElement(root, "student")` |
| `element.text = "текст"` | Добавляет текстовое содержимое в элемент | `name.text = "Ivan"` |
| `element.set(key, value)` | Добавляет атрибут к элементу | `student.set("id", "1")` |
| `element.attrib` | Словарь атрибутов элемента | `print(student.attrib)` |
| `ET.tostring(element)` | Преобразует элемент в строку (байты) | `ET.tostring(root, encoding="unicode")` |
| `ET.ElementTree(element)` | Создает дерево из элемента | `tree = ET.ElementTree(root)` |
| `tree.write(file)` | Записывает дерево в файл | `tree.write("students.xml", encoding="utf-8")` |

ET - ElementTree

```python

''' Создание нового элемента (корневого элемента) '''
root = ElementTree.Element('studentList')

''' Создание дочерних элементов '''
student1 = ElementTree.SubElement(root,'student') # Создаю элемент студент 1
student2 = ElementTree.SubElement(root,'student') # Создаю элемент студент 2


name1 = ElementTree.SubElement(student1,'name') # Создаю имя для студента 1
lastname1 = ElementTree.SubElement(student1, 'lastname') # Создаю фамилию для студента 1
score1 = ElementTree.SubElement(student1, 'score') # Создаю счет для студента 1
gender1 = ElementTree.SubElement(student1, 'gender') # Создаю пол для студента 1

name2 = ElementTree.SubElement(student2,'name') # Создаю имя для студента 2
lastname2 = ElementTree.SubElement(student2, 'lastname') # Создаю фамилию для студента 2
score2 = ElementTree.SubElement(student2, 'score') # Создаю счет для студента 2
gender2 = ElementTree.SubElement(student2, 'gender') # Создаю пол для студента 2

''' Заполнение данными '''

name1.text = 'Утямомо'
lastname1.text = 'Утямомович'
score1.text = '100.0'
gender1.text = 'True'

name2.text = '3e12'
lastname2.text='323'
score2.text = '321321'
gender2.text = 'False'
''' Добавление атрибутов '''

student1.set('id', '1')
student2.set('id', '2')
score1.set('type', 'float')
score2.set('type', 'float')

''' Сохранение в файл '''

tree = ElementTree.ElementTree(root)
tree.write("students.xml", encoding="utf-8", xml_declaration=True)
```

### **Чтение XML**" и **Поиск элементов**

| Команда | Описание | Пример |
|---------|----------|--------|
| `ET.parse(file)` | Читает XML-файл и возвращает ElementTree | `tree = ET.parse("students.xml")` |
| `tree.getroot()` | Возвращает корневой элемент дерева | `root = tree.getroot()` |
| `ET.fromstring(string)` | Парсит XML из строки | `root = ET.fromstring(xml_string)` |
| `element.tag` | Возвращает название тега | `print(student.tag)` |
| `element.attrib` | Возвращает словарь атрибутов | `print(student.attrib)` |
| `element.text` | Возвращает текстовое содержимое | `print(name.text)` |
| `element.get(key)` | Получает значение атрибута по ключу | `student.get("id")` |



| Команда | Описание | Пример |
|---------|----------|--------|
| `element.find(tag)` | Находит первый дочерний элемент с указанным тегом | `student.find("name")` |
| `element.findall(tag)` | Находит все дочерние элементы с указанным тегом | `root.findall("student")` |
| `element.iter(tag)` | Создает итератор по всем элементам (включая вложенные) с указанным тегом | `root.iter("score")` |
| `element.iterfind(match)` | Находит все элементы, соответствующие шаблону | `root.iterfind(".//student")` |


```python
''' Читаем XML-файл '''

tree = ElementTree.parse('students.xml')

''' Выводим корневой каталог '''

root = tree.getroot()
print(f'Корневой каталог:{root}')

''' Нахождение всех дочерних элементов '''

print(root.findall('student'))

''' Пройдемся по всем элементам'''

for student in root.findall("student"): # Найдем всех студентов
    student_id = int(student.get("id")) # Запишем id каждого студента в переменную и преобраазуем его в int
    name = student.find("name").text # Найдем имя и запишем его
    score = float(student.find("score").text) # Найдем счет и запишем его
    print(f"ID: {student_id}, Name: {name}, Score: {score}")
    print('Словарь атрибутов:',student.attrib)

''' Итератор по всем элементам '''

for score in root.iter('score'):
    print(score.text)
```

### **Изменение XML**

| Команда | Описание | Пример |
|---------|----------|--------|
| `element.text = "новый текст"` | Изменяет текст элемента | `name.text = "New Name"` |
| `element.set(key, value)` | Добавляет или изменяет атрибут | `student.set("status", "active")` |
| `element.remove(child)` | Удаляет дочерний элемент | `root.remove(student)` |
| `element.append(child)` | Добавляет дочерний элемент | `root.append(new_student)` |
| `element.extend(iterable)` | Добавляет несколько дочерних элементов | `root.extend([student1, student2])` |
| `element.clear()` | Очищает элемент (удаляет текст, атрибуты, детей) | `student.clear()` |

```python
import copy

''' Создаем копию, чтобы не изменять основной файл '''
tree_copy, root_copy = copy.deepcopy(tree), copy.deepcopy(root)

''' Изменим или добавляем атрибут '''
for student in root_copy.findall('student'):
    print('Изменение атрибутов:')
    print(student.get('id'))
    student.set('id', '0')
    print(student.get('id'))
    print('---' * 20)
```

```python
''' Создаем копию, чтобы не изменять основной файл '''
tree_copy, root_copy = copy.deepcopy(tree), copy.deepcopy(root)

''' Изменим текст '''
for student in root_copy.findall('student'):
    print('Изменение текста:')
    score = student.find('score')
    print('score до изменения:', score.text)
    score.text = 100
    print('score после изменения:', score.text)
    print('---' * 20)
```

```python
''' Создаем копию, чтобы не изменять основной файл '''
tree_copy, root_copy = copy.deepcopy(tree), copy.deepcopy(root)

number_rows = 0 # Номер строки для удобства чтения

''' Удалим дочерний элемент '''
for student in root_copy.findall('student'):
    print(f'Удаление дочернего элемента {number_rows}:')
    name = student.find('name')
    print(student.find('name').text)
    student.remove(name)
    print(student.find('name'))
    print('---' * 20)
    number_rows +=1
```

```python
''' Для простоты запишем все имена'''

names = [name.text for name in root.iter('name')]
number_row = 0

''' Добавление дочернего элемента '''

for student in root_copy.findall("student"):
    if number_row <= len(names):
        subname = ElementTree.SubElement(root,'name') #Создание дочернего элемента
        subname.text = names[number_row]
        student.append(subname)
        number_row+=1
        print(student.find('name').text)
```

```python
''' Создаем копию, чтобы не изменять основной файл '''
tree_copy, root_copy = copy.deepcopy(tree), copy.deepcopy(root)

''' Список всех тегов '''

tags = set([tag_.tag for tag_ in root.iter()]) # Пройдемся по всем возможным значениям и оттуда вытащим тэги
number_row = 0 # Для удобства отображения

'''Очищение элементов'''

for student in root_copy.findall('student'):
  print(f'{"---"*10} Элемент №{number_row} {"---"*10}')
  print('До удаления')
  print(f'Словарь атрибутов: {student.attrib}')
  for tag in tags:
    print(f'тэг {tag}: {student.find(tag)}')

  student.clear()
  print('\nПосле удаления')

  for tag in tags:
    print(f'тэг {tag}: {student.find(tag)}')

  print(f'Словарь атрибутов:{student.attrib}')
  print
  number_row += 1
```
