# **API**

**API** (*Application Programming Interface — программный интерфейс приложения*) — это набор правил и спецификаций, который позволяет разным программам взаимодействовать друг с другом. Но потребности в этом взаимодействии бывают разные, поэтому и появились разные типы API, каждый со своими сильными сторонами и сценариями использования.

https://habr.com/ru/companies/otus/articles/737610/

## **Типы api:**

- **REST API** (Representational State Transfer)
    - **Простыми словами**: Универсальный язык для веб-сервисов 
    - **Как работает**: Использует URL для идентификации ресурсов (например, /users, /products/123) и HTTP-методы (GET, POST, PUT, DELETE) для выполнения действий над этими ресурсами. **Данные обычно передаются в форматах JSON или XML.**
    - **Преимущества**: Простота, Гибкость
    - **Недостатки**: under-fetching (нужно несколько запросов для получения всех данных), over-fetching (получается больше данных чем нужно)
