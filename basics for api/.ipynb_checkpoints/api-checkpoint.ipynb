# **API**

**API** (*Application Programming Interface — программный интерфейс приложения*) — это набор правил и спецификаций, который позволяет разным программам взаимодействовать друг с другом. Но потребности в этом взаимодействии бывают разные, поэтому и появились разные типы API, каждый со своими сильными сторонами и сценариями использования.

https://habr.com/ru/companies/otus/articles/737610/

## **Типы API**

- **REST API** (Representational State Transfer)
    - **Простыми словами**: Универсальный язык для веб-сервисов.
    - **Как работает**: Использует URL для идентификации ресурсов (например, /users, /products/123) и HTTP-методы (GET, POST, PUT, DELETE) для выполнения действий над этими ресурсами. **Данные обычно передаются в форматах JSON или XML.**
    - **Преимущества**: Простота, Гибкость.
    - **Недостатки**: under-fetching (нужно несколько запросов для получения всех данных), over-fetching (получается больше данных чем нужно).
    - **Сценарии использования**: Общее назначение, широкое применение (веб-сервисы, мобильные приложения).

- **SOAP API** (Simple Object Access Protocol)
    - **Простыми словами**: Строгий, официальный протокол для надежных систем, где важна безопасность и транзакции.
    - **Как работает**: SOAP API более структурированы и часто требуют использования WSDL (Web Services Description Language) файла, который описывает, как клиент должен взаимодействовать с сервисом. Обмен сообщениями происходит через HTTP, но может использовать и другие протоколы. Сообщения пишутся в строгом формате XML.
    - **Преимущества**: Высокая безопасность, стандартизация, поддержка сложных транзакций, независимость от транспортного протокола.
    - **Недостатки**: Сложность в разработке и отладке, "тяжеловесность" (XML-сообщения занимают много места), медленнее REST.
    - **Сценарии использования**: Распределенные энтерпрайз среды, унаследованные системы (например, банковские системы, системы бронирования билетов, платежные шлюзы, где критически важна надежность и безопасность).

- **WebSocket API**
    - **Простыми словами**: Постоянная телефонная линия между клиентом и сервером, по которой они могут обмениваться сообщениями в реальном времени.
    - **Как работает**: Клиент отправляет запрос на установку соединения (handshake), после чего сервер и клиент «держат трубку» (постоянный канал открыт). Любая из сторон может инициировать отправку данных в любой момент, не дожидаясь запроса.
    - **Преимущества**: Двусторонняя связь в реальном времени, низкая задержка, эффективность (нет накладных расходов на установку нового соединения для каждого сообщения).
    - **Недостатки**: Сложнее в масштабировании (серверу нужно держать много одновременных соединений), не подходит для обычных CRUD-операций (создание/чтение/обновление/удаление).
    - **Сценарии использования**: Чаты, онлайн-игры, биржевые котировки и торговля, совместное редактирование документов (как в Google Docs), ленты уведомлений.

- **gRPC API** (Google Remote Procedure Call)
    - **Простыми словами**: Высокоскоростной курьер, который упаковывает данные очень плотно и говорит на двоичном языке для максимальной скорости внутри дата-центров.
    - **Как работает**: Использует Protocol Buffers (protobuf) — компактный двоичный формат для данных и строгий контракт в виде `.proto` файла. Клиент вызывает метод на сервере так, как будто это локальная функция, благодаря чему все работает очень быстро.
    - **Преимущества**: Очень высокая производительность и скорость, строгая типизация (меньше ошибок), поддерживает стриминг данных (как и WebSocket), автогенерация кода для многих языков.
    - **Недостатки**: Сложность настройки, двоичный формат (protobuf) не читается человеком в сыром виде (в отличие от JSON), меньше поддержка в браузерах (обычно используется для внутренних сервисов).
    - **Сценарии использования**: Микросервисная архитектура (общение между сервисами внутри компании), высоконагруженные системы, мобильные приложения (где важна экономия трафика и батареи), стриминг видео/данных.

- **GraphQL API**
    - **Простыми словами**: Заказ еды в ресторане, где вы говорите официанту (GraphQL-серверу) точный список блюд (полей), которые хотите получить, и вам приносят только их.
    - **Как работает**: У API есть одна точка входа (один URL). Клиент отправляет на него не просто запрос, а «запрос-спецификацию» (query), где описывает структуру и поля нужных данных. Сервер собирает ответ ровно по этому описанию из разных источников.
    - **Преимущества**: Решает проблемы over-fetching и under-fetching (получаешь только то, что просил), удобен для быстрой разработки фронтенда, позволяет получать данные из разных источников за один запрос.
    - **Недостатки**: Сложность кэширования (сложнее, чем с REST), сложность запросов (клиент может запросить слишком много вложенных данных и положить сервер), дополнительная нагрузка на сервер (парсинг сложных запросов).
    - **Сценарии использования**: Сложные интерфейсы с составными данными (дашборды, профили пользователей), мобильные приложения (где важен каждый килобайт трафика), агрегация данных из нескольких источников (микросервисов) для фронтенда.

## **Таблица из статьи** (https://habr.com/ru/companies/otus/articles/737610/)

| Характеристика | REST API | SOAP API | WebSocket API | gRPC API | GraphQL API |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Формат данных** | Как правило, JSON, но может поддерживать несколько форматов | XML | Обычно JSON, но может быть любой тип данных | Protocol Buffers (бинарный формат) | JSON |
| **Протокол** | HTTP/HTTPS | HTTP/HTTPS, SMTP, XMPP, и другие | WS/WSS (WebSocket secure) | HTTP/2 | HTTP/HTTPS |
| **Тип связи** | Односторонняя связь (запрос/ответ) | Односторонняя связь (запрос/ответ) | Двусторонняя связь (полный дуплекс) | Двунаправленная потоковая передача | Односторонняя связь (запрос/ответ) |
| **Service Discovery** | Не стандартизированы, часто используют OpenAPI (Swagger) для документации | Стандартизация с помощью WSDL | Не стандартизировано | Не стандартизирован, но поддерживает Server Reflection | Не стандартизировано, интроспекция для обнаружения схем |
| **Производительность** | Варьируется, обычно ниже, чем бинарные протоколы из-за текстовой природы | Снижена из-за словесного формата XML | Выше благодаря полнодуплексной связи и отсутствию накладных расходов HTTP | Высокий (бинарный протокол, преимущества HTTP/2) | Варьируется, эффективная загрузка данных может повысить производительность |
| **Сценарий использования** | Общее назначение, широкое применение | Распределенные энтерпрайз-среды, унаследованные системы | Приложения в режиме реального времени, когда серверу необходимо пушить обновления | Микросервисы, системы, критичные к производительности | Когда требуется гибкая и эффективная загрузка данных |
| **Сложность** | От низкого до умеренного | Высокая | Умеренная | От умеренной до высокой | Умеренная |

# Примеры использования API (самого простого)

```python
import requests
import json

api = '11c0d3dc6093f7442898ee49d2430d20'
url = 'http://api.openweathermap.org/geo/1.0/direct?'

params = {
    'q' : 'Saint-Petersburg',
    'appid' : api
}

result = requests.get(url, params=params)

json.loads(result.text)[0]['lat']
```
