```python
from lxml import etree
import requests

url = 'https://steamcommunity.com/market/search?appid=730#p1_popular_desc'
steam = requests.get(url)
parser = etree.HTMLParser()

root = etree.fromstring(steam.text, parser)


for element in root.xpath('//span[@class=\'market_listing_item_name\']'):
  print(f'{element.text}')

for element in root.xpath('//span[@class=\'normal_price\']'):
  print(f'{element.attrib} --- {element.text}')
```
