# **Запросы (различные команды) Request**

## **Основные команды: (часто используемые)**

- **GET** — получить данные (Запрашивает ресурс с сервера, **не изменяет состояние на сервере**), если ресурс является html-страницей, то мы получим ответ в виде кода этой страницы.
    - ***Когда применять***: Получить страницу, список объектов, данные по id. 
- **POST** — отправить/создать данные (Отправляет данные на сервер, **обычно создает новый ресурс или выполняет действие**)
    - ***Когда применять***: Создать новый объект (например, регистрацию, добавление записи).
## **Редко используемые команды:**
- **PUT** — заменить ресурс полностью (Отправляет полную версию ресурса; **сервер обычно заменяет существующий ресурс этой версией.**)
    - ***Когда применять***: Полная замена объекта (все поля).
- **PATCH** — частично обновить ресурс (**Отправляет изменения только для некоторых полей ресурса.**)
    - ***Когда применять***: Обновить одно или несколько полей, не трогая остальные.
- **DELETE** — удалить ресурс (**Запрашивает удаление ресурса на сервере.**)
    - ***Когда применять***: Удалить запись по id.
- **HEAD** — получить только заголовки (**Как GET, но возвращает только заголовки ответа, без тела.**)
    - ***Когда применять***: Проверить существование ресурса, размер, тип контента, кеширование.

## **Статус коды:** 
Основные классы кодов HTTP

- **1xx: *Informational (Информационные)*** — запрос получен, обработка продолжается.
- **2xx: *Success (Успех)*** — запрос успешно получен, понят и принят.
- **3xx: *Redirection (Перенаправление))*** — требуются дополнительные действия клиента.
- **4xx: *Client Error (Ошибка клиента))*** — запрос содержит ошибки или не может быть выполнен.
- **5xx: *Server Error (Ошибка сервера))*** — сервер не может выполнить запрос. 

## **Примеры:**

### **Для HTML:**

```python
import requests

''' Посмотреть статус код'''
result = requests.get('https://docs.python.org/3/')
print(f'{result.status_code}')
```

```python
''' Посмотреть тип содержимого, также можно посмотреть во сколько был сделан запрос и ещё много всякой информации'''

print(f'{result.headers['Content-Type']}\n{'---'*15}')

''' Просмотр различной информации'''

print(result.headers)
```

```python
''' Просмотр содержимого'''

print(result.content[:200])

print(result.text[:200])
```

```python
''' Также можно задать имя параметра'''

result = requests.get('https://ya.ru/search/',params={'text': '321'})
result.text[:200]
```

### **Для картинок:**

```python
pic = requests.get('https://shared.fastly.steamstatic.com/store_item_assets/steam/apps/2139460/ss_78c7278d10652c5369891393782d84b6934176c5.600x338.jpg')

print(pic.status_code)

print(pic.headers['Content-Type'])

print(pic.content[:200])

''' Чтобы увидеть картинку, мы можем её просто сохранить в файле'''

with open('files/pic.jpg', 'wb') as p:
    p.write(pic.content)
```
