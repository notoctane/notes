# **Работа с таблицами**

## **SELECT**

Данный оператор используется для извлечения данных из одной или нескольких таблиц.

Самый простой синтаксис данного оператора имеет вид:

`[DISTINCT]` - необязательный параметр. Удаляет дубликаты из набора результата.

<список_выражений> -  столбцы или вычисления, которые вы хотите получить. Используйте *, если вы хотите выбрать все столбцы.

<таблица> - таблицы, из которых вы хотите получить записи. Должна быть хотя бы одна таблица, перечисленная в предложении FROM

`[WHERE <условие>]` - необязательный параметр. Условия, которые должны быть выполнены для выбранных записей.

`[GROUP BY <список_группировки>]` - необязательный параметр. Он собирает данные по нескольким записям и группирует результаты по одному или нескольким столбцам.

`[HAVING <условие>]` - необязательный параметр. Он используется в сочетании с GROUP BY, чтобы ограничить группы возвращаемых строк только теми, чье условие Правда. 

`[ORDER BY <список сортировки>[ ASC | DESC ]]` - необязательный параметр. Он используется для сортировки записей в вашем результирующем наборе. ASC - сортировка по возрастанию, DESC  - сортировка по убыванию.

`[LIMIT <количество_строк>]` -  задает ограниченное количество строк в результирующем наборе. Например, LIMIT 10 вернет первые 10 строк, соответствующих критериям SELECT. В этом случае порядок сортировки имеет значение, поэтому обязательно используйте оператор ORDER BY соответствующим образом.

## **Логические операции** (`AND` (`&&`), `OR` (`||`), `NOT`)

Условия состоят из операторов сравнения (перечислены на следующем шаге) и операторов `NOT` (НЕ), `AND`(И) и `OR`(ИЛИ). **Приоритет `NOT` выше, чем `AND`, а приоритет `AND` выше, чем `OR`**. Для указания последовательности вычисления условий можно использовать круглые скобки.

### **Список операторов**

| Операторы сравнения | Описание                                | Пример                                                          |
|---------------------|-----------------------------------------|-----------------------------------------------------------------|
| =                   | Равно                                   | where cabinet_num = 21                                          |
| <>!                 | Не равно                                | where cabinet_num != 21                                         |
| >                   | Больше, чем                             | where cabinet_num > 21                                          |
| >=                  | Больше или равно                        | where cabinet_num >= 21                                         |
| <                   | Меньше, чем                             | where cabinet_num < 21                                          |
| <=                  | Меньше или равно                        | where cabinet_num <= 21                                         |
| IN()                | Соответствует значению в списке         | where cabinet_num in (21, 11)                                   |
| BETWEEN             | В пределах диапазона (включительно)     | where cabinet_num between 10 and 20                             |
| IS NULL             | Значение NULL                           | where cabinet_num IS NULL                                       |
| IS NOT NULL         | Значение, не NULL                       | where cabinet_num IS NOT NULL                                   |
| LIKE                | Соответствие шаблону % и _              | where full_name like 'Иван%' (ФИО начинается на Иван)           |

### **NULL**

`NULL` обозначает отсутствие данных или неопределенное значение. **`NULL` это не то же самое, что 0 и пустая строка**.

`NULL` при сравнении (=, <, >) с другими значениями всегда выдает в результате NULL, поэтому **для сравнения с `NULL` всегда используйте конструкцию `IS NULL`.**

## **Встроенные функции**

| Функция | Описание | Пример |
| :--- | :--- | :--- |
| **CONCAT(expr1, expr2, ..., exprN)** | Позволяет объединять вместе два или более выражений | `SELECT CONCAT('П', 'р', 'и ', 'а', 'е', 'т') AS ex_concat;`<br>**Результат:** ex_concat = Привет! |
| **LOWER(строка)** | Преобразует все символы указанной строки в строчные (нижний регистр) | `SELECT LOWER('ПриВер1234 ') AS ex_lower;`<br>**Результат:** ex_lower = Привет!1234 |
| **UPPER(строка)** | Преобразует все символы указанной строки в верхний регистр | `SELECT UPPER('ПриВер1234 ') AS ex_upper;`<br>**Результат:** ex_upper = Привет!1234 |
| **ABS(число)** | Возвращает абсолютное значение числа | `SELECT ABS(-12.36) AS ex_abs;`<br>**Результат:** ex_abs = 12.36 |
| **GREATEST(n1, n2, ..., n)** | Возвращает наибольшее значение в списке выражений | `SELECT GREATEST(8,9,23,4,5,78,2,13,75) AS ex_greatest;`<br>**Результат:** ex_greatest = 78 |
| **LEAST(n1, n2, ..., n)** | Возвращает наименьшее значение в списке выражений | `SELECT LEAST(8,9,23,45,7,8,2,13,75) AS ex_least;`<br>**Результат:** ex_least = 2 |
| **ROUND(число, [число знаков])** | Возвращает число, округленное до определенного количества десятичных знаков | `SELECT ROUND(8.589) AS ex_round1, ROUND(8.589, 2) AS ex_round2;`<br>**Результат:** ex_round1 = 8.59, ex_round2 = 8.59 |
| **CURDATE()** | Возвращает текущую дату | `SELECT CURDATE() AS ex_curdate;`<br>**Результат:** `ex_curdate = 2024-01-14` |
| **DAY(дата)** | Возвращает номер дня месяца из даты | `SELECT DAY('2024-01-14') AS ex_day;`<br>**Результат:** `ex_day = 14` |
| **MONTH(дата)** | Возвращает числовое значение месяца из даты | `SELECT MONTH('2024-01-14') AS ex_month;`<br>**Результат:** `ex_month = 1` |
| **DATEDIFF(дата1, дата2)** | Возвращает число дней между двумя датами. Если `дата1 > дата2` — положительное, иначе — отрицательное | `SELECT DATEDIFF('2024-01-14', '2023-11-12') AS ex_datediff;`<br>**Результат:** `ex_datediff = 63` |
| **YEAR(дата)** | Возвращает год из значения даты | `SELECT YEAR('2024-01-14') AS ex_year;`<br>**Результат:** `ex_year = 2024` |
| **ISNULL(выражение)** | Проверяет, является ли выражение NULL (возвращает 1, если NULL, иначе 0) | `SELECT ISNULL('') AS ex_isnull1, ISNULL(NULL) AS ex_isnull2, ISNULL('Привет') AS ex_isnull3;`<br>**Результат:** `ex_isnull1 = 0`, `ex_isnull2 = 1`, `ex_isnull3 = 0` |
| **IF(условие, знач_если_True, знач_если_False)** | Возвращает одно значение, если условие истинно (TRUE), и другое, если ложно (FALSE) | `SELECT IF(50<100, 'y', 'n') AS ex_if;`<br>**Результат:** `ex_if = y` |
| **COALESCE(выр1, выр2, ..., вырN)** | Возвращает первое не NULL выражение в списке | `SELECT COALESCE(NULL, NULL, 'Привет', NULL, 'Пока') AS ex_coalesce;`<br>**Результат:** `ex_coalesce = Привет` |

# **Агрегатные функции**

Отличия агрегатных функций от скалярных в том, что:
- **скалярные** функции выполняют вычисления над **одним значением или списком значений**
- **агрегатные** функции выполняют вычисления над **группой значений из нескольких строк**

*Проще говоря, в рамках работы с таблицей, скалярные функции выполняют вычисления в рамках одной строки, а агрегатные функции выполняют вычисления в рамках нескольких строк.*

| Название функции | Описание |
|------------------|----------|
| **AVG(выражение)**   | Вычисляет среднее значение |
| **SUM(выражение)**   | Вычисляет сумму значений |
| **MIN(выражение)**   | Вычисляет наименьшее значение |
| **MAX(выражение)**   | Вычисляет наибольшее значение |
| **COUNT(выражение)** | Вычисляет количество значений в указанном выражении `(Считает NULL)` |
| **COUNT(*)**         | Возвращает количество строк источника записей |

# **GROUPBY, HAVING**

## **GROUP BY**

Поля, по которым группируются строки, должны быть перечислены после оператора GROUP BY. Все поля, не входящие в список группировки, должны вычисляться при помощи агрегатных функций.
