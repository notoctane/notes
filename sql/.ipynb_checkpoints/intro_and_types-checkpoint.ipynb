# **SQL (Structured Query Language)**

## **Общие термины**

- **SQL** -- это язык структурированных запросов и используют его для работы с базами данных(БД)
- **Реляционная база данных** – это набор данных с предопределенными связями(отношениями) между ними. В реляционной БД данные представляются в виде плоской таблицы.
- Каждая строка таблицы представляет собой набор связанных значений, относящихся к одному объекту или сущности. Каждая строка в таблице может быть помечена уникальным идентификатором, называемым **первичным ключом (Primary Key)**, а строки из нескольких таблиц могут быть связаны с помощью внешних ключей (Foreign key).
**Требования к первичному ключу:**

    - **Уникальность**: Каждое значение в столбце (или комбинации столбцов) должно быть уникальным.

    - **Отсутствие NULL**: Значения в столбце не могут быть пустыми (NULL).

    - **Неизменяемость**: Значения первичного ключа не должны изменяться со временем (или изменяться крайне редко).

- **Система управления базами данных (СУБД)** — это программа, которая помогает работать с данными, представленными в виде таблиц. Это программа помогает разложить информацию таким образом, чтобы было легче её искать и работать с ней.
    - СУБД собирает данные в специальные таблицы, чтобы они не были разбросаны.
    - Она делит данные на категории (например, пользователи, товары, заказы), чтобы было проще находить нужную информацию.
    - СУБД помогает быстро искать данные. Если тебе нужно найти номер телефона друга, ты просто спрашиваешь у СУБД, и она быстро находит его для тебя


## **Типы данных:**

### **Строка**, бывает нескольких видов:

- **Строка фиксированной длины** (`CHAR(n)`, n - кол-во символов в строке). **Максимальная длина строки (`n`) --255.** **Память всегда выделяется под n-символов для всего датасета**, даже в случаях когда слово меньшего числа символов. **Используется в случаях**:
    - Когда все строки должны быть одной длины (например, коды товаров, индексы, ID).
    - При работе со старым ПО, фиксированными форматами данных, где важна длина.
    - Иногда чуть быстрее работает сравнение, потому что СУБД знает точную длин

**Пример**


![изображение.png](image/intro_and_types/char.png) 

---

- **Обычная строка** (`VARCHAR(n)`, где n -- максимальное число символов в поле). Максимальный размер (`n`) -- 65535. Память выделяется только под количество символов в строке в каждой строке датасета.

**Пример:**


![изображение.png](image/intro_and_types/varchar.png) 

--- 

- **Строка большого размера** (`TEXT`). Существует 4 типа `TINNYTEXT`, `TEXT`, `MEDIUMTEXT`, `LONGTEXT`, они отличаются только по максимально длинне значений.

- `ENUM` строковый объект со значением, выбранным из списка разрешенных значений, которые перечислены явно в спецификации столбца при создании таблицы. Позволяет **хранить одно значение из заранее определенного списка**. **Например, ENUM('красный', 'желтый', 'зеленый') позволяет выбрать только одно из этих значений**

- `SET` строковый объект, у которого может быть ноль или больше значений, каждое из которых должно быть выбрано из списка разрешенных значений, определенных, когда таблица составлена. Позволяет **хранить несколько значений из заданного списка**. **Например, SET('груша', 'апельсин', 'банан') может содержать любое сочетание этих значений, включая все сразу**\

---

### **Бинарный тип:**

- **Двоичная строка** (*binary strings*) - существует для хранения последовательности байтов. Также как и для строк существуют 2 типа: `VARBINARY(n)` и `BINARY(n)`. **Что можно хранить в этих строках:**
    - **Изображения**
    - **Аудио-видео данные**
    - **Хэши паролей**
    - Сжатые данные и тд. (Любая информация которую можно представить последовательности байтов)
  Как для строк существует строка с увеличеным максимальным количеством длины значений (`TINNYBLOB`, `BLOB`, `MEDIUMBLOB`, `LONGBLOB`). Обычно в нем хранятся картинки, видео.

### **Логические типы**:

- **Булево число** (`BOOLEAN`). True or False.

### **Числовые типы:**

- **Типы с фиксированной точкой** (`DECIMAL(<точность>,<масштаб>)`.
    - **<точность>** - общее количество разрядов. Если точность не указана, то по умолчанию это будет 10. Максимальное значение - 65.
    - **<масштаб>** - количество разрядов справа от десятичной точки. Значение по умолчанию для масштаба равно 0.

    Данный тип также имеет псевдонимы NUMERIC, DEC, FIXED. **Примеры использования**:


**Пример 1**: `DECIMAL(5)`

| Значение         | Как будет храниться в столбце |
|------------------|-------------------------------|
| **12345**        | 12345                         |
| **123456**       | Ошибка при записи в столбец (превышено количество разрядов) |
| **123.45**       | 123 (произойдет округление до целого числа) |
| **12**           | 12                            |

---

**Пример 2**: `DECIMAL(5,2)`
  
| Значение         | Как будет храниться в столбце |
|------------------|-------------------------------|
| **123.45**       | 123.45                        |
| **123.456**      | 123.46 (происходит округление до 2 знаков после запятой) |
| **1234**         | Ошибка при записи в столбец (число 1234.00 имеет в сумме больше 5 десятичных разрядов) |
| **1.2**          | 1.2                           |

---

#### **Типы с плавающей точкой:**

 - `FLOAT(M,D)`**,

    (Хранит дробные числа с плавающей точкой одинарной точности от −3.4028 ∗ 10 ^ 38 до 3.4028∗10^38 занимает 4 байта

- `DOUBLE(M,D)`**,

    (Хранит дробные числа с плавающей точкой двойной точности от  −1.7976∗10^308 до 1.7976∗10^308, занимает 8 байт)
---

**где M - общее кол-во цифр, D - кол-во цифр после запятой.

### **Временной тип**

Все типы:

| Дата/время      | Пример                  |
|-----------------|-------------------------|
| **DATE**        | 2023-12-01              |
| **DATETIME**    | 2023-12-01 13:15:18     |
| **TIMESTAMP**   | 2023-12-01 13:15:18     |

| **YEAR**        | 2023                    |

#### **DATE**


Тип `DATE` используется для значений с частью даты, но без части времени. MySQL получает и выводит на экран значения `DATE` в формате 'YYYY-MM-DD'.

**Поддерживаемый диапазон от '1000-01-01' до '9999-12-31'.**

##### **Форматы `DATE`**

| Формат строки (в кавычках)/числа | Разделитель    | Пример               |
|----------------------------------|----------------|----------------------|
| 'YYYY-MM-DD'                     | Любой символ пунктуации | ‘2023-01-12’     |
|                                  |                | ‘2023@01@12’     |
| 'YY-MM-DD'                       | Любой символ пунктуации | ‘23-01-12’       |
|                                  |                | ‘23@01@12’       |
| 'YYYYMMDD'                       | -              | ‘20230112’           |
| 'YYMMDD'                         | -              | ‘230112’             |
| YYYYMMDD                         | -              | 20230112             |
| YYMMDD                           | -              | 230112               |


---

#### **DATETIME**

Тип `DATETIME` используется для значений с частями даты и времени. MySQL получает и выводит на экран значения DATETIME в формате 'YYYY-MM-DD HH:MI:SS'.

**Поддерживаемый диапазон от '1000-01-01 00:00:00' до '9999-12-31 23:59:59'.**

**Этот тип данных не зависит от временной зоны**, установленной в MySQL. **Он всегда отображается ровно в таком виде, в котором был установлен и в котором хранится в базе данных**. То есть при изменении часового пояса, отображение времени не изменится.

#### **TIMESTAMP**

Тип `TIMESTAMP` используется для значений, которые содержат части даты и времени. 

**TIMESTAMP имеет диапазон от '1970-01-01 00:00:01' UTC до '2038-01-19 03:14:07' UTC.**

**При выборках отображается с учётом текущего часового пояса**. Часовой пояс можно задать в настройках операционной системы, где работает MySQL, в глобальных настройках MySQL или в конкретной сессии. **В базе данных при создании записи с типом TIMESTAMP значение сохраняется по нулевому часовому поясу**. Это основное отличие от предыдущего типа `DATETIME`.

#### **Форматы `DATETIME` и `TIMESTAMP`**

| Формат строки (в кавычках)/числа | Разделитель    | Пример                   |
|----------------------------------|----------------|--------------------------|
| 'YYYY-MM-DD HH:MI:SS'            | Любой символ пунктуации | '2023-01-12 13:14:35' |
| 'YY-MM-DD HH:MI:SS'              | Любой символ пунктуации | '23-01-12 13:14:35'   |
| 'YYYYMMDDHHMMISS'                | -              | '20230112131435'         |
| 'YYMMDDHHMMISS'                  | -              | '230112131435'           |
| YYYYMMDDHHMMISS                  | -              | 20230112131435           |
| YYMMDDHHMMISS                    | -              | 230112131435             |

#### **TIME**

MySQL получает и отображает значения `TIME` в формате 'HH:MI:SS' (или 'HHH:MI:SS' для больших значений часов). 

**Значения TIME могут быть в диапазоне от '-838:59:59' до '838:59:59'.**

##### **Форматы TIME**

| Формат строки                 | Пример         |
|-------------------------------|----------------|
| 'D HH:MI:SS'                  | '4 13:45:14'   |
| 'HH:MI:SS'                    | '13:45:14'     |
| 'HH:MI'                       | '13:45'        |
| 'D HH:MI'                     | '4 13:45'      |
| 'D HH'                        | '4 13'         |
| 'SS'                          | '14'           |
| **Данные в скобках**          |                |
| *D – день, от 0 до 34.*       |                |
| 'HHMISS'                      | '134514'       |
| HHMISS                        | 134514         |
| SS                            | 14             |
| MISS                          | 4514           |

#### **YEAR**

Тип `YEAR` 1-байтовый тип, используемый, чтобы представить значения года. 

**MySQL отображает YEAR в формате YYYY с диапазоном от 1901 до 2155 или 0000.**

##### **Форматы YEAR**

| Строка/число                   | Диапазон             | Пример    |
|--------------------------------|----------------------|-----------|
| Число с 4 цифрами              | от 1901 до 2155      | 1987      |
| Строка с 4 цифрами             | от '1901' до '2155'  | ‘1987’    |
| Число с 1 или 2 цифрами        | от 1 до 99           | 87        |
| Строка с 1 или 2 цифрами       | от ‘1’ до ’99’       | ‘87’      |
