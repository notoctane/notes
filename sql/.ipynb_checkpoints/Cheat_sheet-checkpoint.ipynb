| Тип данных | Подтип | Максимальный размер | Особенности хранения | Применение / Примечания |
| :--- | :--- | :--- | :--- | :--- |
| **Строковые** | `CHAR(n)` | n ≤ 255 символов | Память всегда выделяется под `n` символов, даже если значение короче (дополняется пробелами). | Коды товаров, индексы, ID. Чуть быстрее при сравнении, так как длина известна заранее. |
| | `VARCHAR(n)` | n ≤ 65 535 символов | Память выделяется только под фактическую длину строки (+1-2 байта на хранение длины). | Основной тип для хранения строк переменной длины (имена, описания). |
| | `TINYTEXT` | 255 байт | Хранит текст вне строки таблицы (в области BLOB). | Для коротких заметок или комментариев. |
| | `TEXT` | 65 535 байт (64 КБ) | Хранит текст вне строки таблицы. | Для статей, постов, описаний среднего размера. |
| | `MEDIUMTEXT` | 16 777 215 байт (16 МБ) | Хранит текст вне строки таблицы. | Для больших текстовых документов, книг. |
| | `LONGTEXT` | 4 294 967 295 байт (4 ГБ) | Хранит текст вне строки таблицы. | Максимально возможный объем символьных данных. |
| | `ENUM` | 65 535 различных значений | Фактически хранит индекс (число), а не строку. | Хранение **одного** значения из списка: `ENUM('красный', 'желтый', 'зеленый')`. |
| | `SET` | 64 элемента в наборе | Фактически хранит битовую маску. | Хранение **комбинации** значений: `SET('груша', 'апельсин', 'банан')`. |
| **Бинарные** | `BINARY(n)` | n ≤ 255 байт | Фиксированная длина, дополняется нулевыми байтами `0x00`. | Хранение хэшей (MD5), бинарных флагов, ключей фиксированной длины. |
| | `VARBINARY(n)` | n ≤ 65 535 байт | Переменная длина, хранится фактическое количество байт. | Сырые бинарные данные переменной длины. |
| | `TINYBLOB` | 255 байт | Бинарный аналог `TINYTEXT`. | Маленькие иконки, сжатые данные. |
| | `BLOB` | 65 535 байт (64 КБ) | Бинарный аналог `TEXT`. | Изображения, аудио/видео файлы небольшого размера. |
| | `MEDIUMBLOB` | 16 777 215 байт (16 МБ) | Бинарный аналог `MEDIUMTEXT`. | Архивы, документы PDF, видеофрагменты. |
| | `LONGBLOB` | 4 294 967 295 байт (4 ГБ) | Бинарный аналог `LONGTEXT`. | Большие мультимедийные файлы, образы дисков. |
| **Логический** | `BOOLEAN` | 1 байт (обычно) | Синоним `TINYINT(1)`. | `TRUE` (1) или `FALSE` (0). Флаги состояния (вкл/выкл). |
