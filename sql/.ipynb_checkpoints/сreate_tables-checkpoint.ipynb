# **Создание таблиц**

## **Синтаксис создания**

Команда для создания:

- **название_таблицы** - это обязательный атрибут.

- **название_столбца** и **тип_данных** - это тоже обязательные атрибуты. Название придумывает разработчик, тип данных разработчик выбирает из существующих в MySQL.

- **атрибуты_столбца** и **атрибуты_таблицы** - их указывать необязательно. Какие значения могут принимать эти атрибуты и что означать - мы разберем на следующих шагах.

## **"Правила" написания скриптов на SQL**

### 1. Форматирование и читаемость
- **Используйте заглавные буквы для ключевых слов SQL** (`SELECT`, `FROM`, `WHERE`, `JOIN`, `INSERT`). Названия таблиц и колонок пишите в нижнем регистре (или в `snake_case`).
- **Разбивайте длинные запросы на строки**. Каждую основную секцию (`SELECT`, `FROM`, `JOIN`, `WHERE`, `GROUP BY`, `ORDER BY`) начинайте с новой строки.
- **Используйте отступы** для подзапросов и условий в `JOIN`/`WHERE`.
- **Явно указывайте схему (префикс)**: `SELECT * FROM dbo.Users` вместо `SELECT * FROM Users`. Это ускоряет компиляцию плана запроса.
- **Используйте `AS` для алиасов** колонок и таблиц (это повышает читаемость).

### 2. Именование объектов
- **Используйте осмысленные имена на английском языке** (единый стандарт в проекте). Избегайте транслита.
- **Таблицы**: существительное во множественном числе (`users`, `orders`, `products`).
- **Первичные ключи**: `id` или `[имя_таблицы]_id`.
- **Внешние ключи**: `[имя_таблицы_на_которую_ссылаемся]_id`.
- **Индексы**: префикс `ix_` + имя таблицы + имена колонок (например, `ix_users_email`).

### 3. Производительность и оптимизация (Критически важно)
- **НИКОГДА не используйте `SELECT *` в продуктовом коде**. Всегда перечисляйте только те колонки, которые реально нужны. `SELECT *` ломает логику при добавлении колонок в таблицу и жрет лишнюю память.
- **Избегайте функций в `WHERE` на индексированных полях** (`WHERE YEAR(Date) = 2025` — плохо. Лучше `WHERE Date >= '2025-01-01' AND Date < '2026-01-01'`).
- **Неявное преобразование типов убивает индексы**. Сравнивайте строку со строкой, число с числом (`WHERE Id = '123'` — плохо, если `Id` числовой).
- **Используйте `EXISTS` вместо `IN` для подзапросов** с большой выборкой данных (почти всегда быстрее).
- **Сначала фильтруйте, потом соединяйте (`JOIN`)**. Старайтесь уменьшить количество строк до выполнения `JOIN`.

### 4. Безопасность и SQL-инъекции
- **Никакой конкатенации строк из пользовательского ввода**. Всегда используйте **параметризованные запросы** (переменные `@param`).
- **Избегайте динамического SQL там, где это возможно**. Если динамический SQL необходим — всегда проверяйте и экранируйте имена объектов через `QUOTENAME()`.
- **Придерживайтесь принципа наименьших привилегий**. Скрипт не должен работать под `sa` или `root`, если ему нужны права только на `SELECT`.

### 5. Транзакции и обработка ошибок
- **Все операции изменения данных (`INSERT`, `UPDATE`, `DELETE`) оборачивайте в транзакцию**, особенно если меняется несколько таблиц.
- **Используйте шаблон `TRY...CATCH`** (в T‑SQL) или аналог `BEGIN ... EXCEPTION` (в PostgreSQL) для отката транзакции при ошибке.
- **Завершайте скрипт явным `COMMIT` или `ROLLBACK`**. Не оставляйте транзакции открытыми.

### 6. Обработка `NULL`
- **Не сравнивайте с `NULL` через `=` или `<>`**. Используйте `IS NULL` или `IS NOT NULL`.
- **Используйте `COALESCE` или `ISNULL` для значений по умолчанию** при вычислениях, чтобы избежать распространения `NULL` на всю строку результата.

### 7. Комментарии и документация
- **Заголовок скрипта**: Обязательно указывайте цель скрипта, автора, дату и версию.
- **Комментируйте сложную бизнес-логику**, а не очевидные вещи (`-- Фильтруем по активным пользователям`, а не `-- Выбираем ID`).
- **Пометка "идемпотентности"**: Если скрипт можно запустить повторно без последствий (например, `CREATE OR ALTER`), укажите это в шапке.

### 8. Работа с JOIN
- **Явно указывайте тип `JOIN`**. Не используйте неявный синтаксис через запятую `FROM a, b WHERE a.id = b.id`. Пишите `FROM a INNER JOIN b ON ...`.
- **Следите за направлением соединений**. Если нужны все записи из левой таблицы, используйте `LEFT JOIN`, а не фильтруйте потом `INNER JOIN`.

### 9. Чек-лист "Безопасный запуск на проде"
- **Всегда начинайте `DELETE` или `UPDATE` с `SELECT`**. Сначала напишите запрос с `SELECT *`, чтобы убедиться, что вы меняете именно те записи, которые планировали.
- **Используйте `TOP` / `LIMIT` / `WHERE` с первичным ключом** при первоначальном тестировании тяжелых изменений.
- **Делайте `BEGIN TRAN`**, проверяйте результат, потом `COMMIT`.

## **Уникальность значения (`UNIQUE`)**

**Согласно правилу `UNIQUE`, в таблице не может существовать двух строк с одинаковыми значениями столбца или набора столбцов**. **При этом неопределенные значения не считаются одинаковыми**, то есть две или более строк со значениями NULL могут быть у столбца/столбцов, заявленные как UNIQUE.

## **Первичный ключ (`PRIMARY KEY`)**

**Первичный ключ** — поле или комбинация полей в таблице, которое позволяет однозначно идентифицировать каждую запись в ней. 

### **Главные свойства `PRIMARY KEY`:**

- **Уникальность** — в таблице не может быть двух строк с одинаковым значением первичного ключа.

- **Непустота** — значение всегда должно быть указано (NOT NULL).

- **Один на таблицу** — у таблицы может быть только один первичный ключ, хотя он может состоять из нескольких столбцов (составной ключ).

## **AUTO_INCREMENT**

**Атрибут столбца. Указывает, что значение данной колонки будет автоматически увеличиваться при добавлении новых записей в таблицу**. Каждая таблица имеет максимум одну `AUTO_INCREMENT` колонку. 

Стоит отметить, что данный параметр можно применять только к целочисленным типам и к типам с плавающей запятой. Последовательности AUTO_INCREMENT начинаются с 1, с шагом 1.
