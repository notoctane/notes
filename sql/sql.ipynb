# **SQL (Structured Query Language)**

## **Общие термины**

- **SQL** -- это язык структурированных запросов и используют его для работы с базами данных (БД)
- **Реляционная база данных** – это набор данных с предопределенными связями (отношениями) между ними. В реляционной БД данные представляются в виде плоской таблицы.
- Каждая строка таблицы представляет собой набор связанных значений, относящихся к одному объекту или сущности. Каждая строка в таблице может быть помечена уникальным идентификатором, называемым **первичным ключом (Primary Key)**, а строки из нескольких таблиц могут быть связаны с помощью внешних ключей (Foreign key).
**Требования к первичному ключу:**

    - **Уникальность**: Каждое значение в столбце (или комбинации столбцов) должно быть уникальным.

    - **Отсутствие NULL**: Значения в столбце не могут быть пустыми (NULL).

    - **Неизменяемость**: Значения первичного ключа не должны изменяться со временем (или изменяться крайне редко).

- **Система управления базами данных (СУБД)** — это программа, которая помогает работать с данными, представленными в виде таблиц. Это программа помогает разложить информацию таким образом, чтобы было легче её искать и работать с ней.
    - СУБД собирает данные в специальные таблицы, чтобы они не были разбросаны.
    - Она делит данные на категории (например, пользователи, товары, заказы), чтобы было проще находить нужную информацию.
    - СУБД помогает быстро искать данные. Если тебе нужно найти номер телефона друга, ты просто спрашиваешь у СУБД, и она быстро находит его для тебя


## **Типы данных:**

- **Строка**, бывает нескольких видов:
  - **Строка фиксированной длины** (`CHAR(n)`, n - кол-во символов в строке). **Максимальная длина строки (`n`) --255.** **Память всегда выделяется под n-символов для всего датасета**, даже в случаях когда слово меньшего числа символов. **Используется в случаях**:
    - Когда все строки должны быть одной длины (например, коды товаров, индексы, ID).
    - При работе со старым ПО, фиксированными форматами данных, где важна длина.
    - Иногда чуть быстрее работает сравнение, потому что СУБД знает точную длин
    
![изображение.png](image/1.png)

  - **Обычная строка** (`VARCHAR(n)`, где n -- максимальное число символов в поле). Максимальный размер (`n`) -- 65535. Память выделяется только под количество символов в строке в каждой строке датасета.

![изображение.png](image/2.png)

  - **Строка большого размера** (`TEXT`). Существует 4 типа `TINNYTEXT`, `TEXT`, `MEDIUMTEXT`, `LONGTEXT`, они отличаются только по максимально длинне значений.
  - `ENUM` строковый объект со значением, выбранным из списка разрешенных значений, которые перечислены явно в спецификации столбца при создании таблицы. Позволяет **хранить одно значение из заранее определенного списка**. **Например, ENUM('красный', 'желтый', 'зеленый') позволяет выбрать только одно из этих значений**

  - `SET` строковый объект, у которого может быть ноль или больше значений, каждое из которых должно быть выбрано из списка разрешенных значений, определенных, когда таблица составлена. Позволяет **хранить несколько значений из заданного списка**. **Например, SET('груша', 'апельсин', 'банан') может содержать любое сочетание этих значений, включая все сразу**

- **Бинарный тип:**
  - **Двоичная строка** (*binary strings*) - существует для хранения последовательности байтов. Также как и для строк существуют 2 типа: `VARBINARY(n)` и `BINARY(n)`. **Что можно хранить в этих строках:**
    - **Изображения**
    - **Аудио-видео данные**
    - **Хэши паролей**
    - Сжатые данные и тд. (Любая информация которую можно представить последовательности байтов)
  Как для строк существует строка с увеличеным максимальным количеством длины значений (`TINNYBLOB`, `BLOB`, `MEDIUMBLOB`, `LONGBLOB`). Обычно в нем хранятся картинки, видео.
